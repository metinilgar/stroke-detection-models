```python
import os
import numpy as np
import nibabel as nib
import imageio

def read_image_series(folder_path):
    """
    Belirli bir klasördeki sıralı görüntüleri okuyarak bir 3D numpy array oluşturur.
    """
    files = sorted([f for f in os.listdir(folder_path) if f.endswith(('.png', '.jpg', '.jpeg'))])
    volume = []
    for file in files:
        img = imageio.imread(os.path.join(folder_path, file))
        volume.append(img)
    return np.stack(volume, axis=-1)  # (H, W, D)

def convert_to_nii_gz(image_dir, mask_dir, output_image_dir, output_mask_dir):
    """
    Image ve mask klasörlerindeki her bir hasta klasörünü okuyarak .nii.gz formatına çevirir.
    """
    patient_ids = sorted(os.listdir(image_dir))
    
    for patient_id in patient_ids:
        image_path = os.path.join(image_dir, patient_id)
        mask_path = os.path.join(mask_dir, patient_id)
        
        if not os.path.isdir(image_path) or not os.path.isdir(mask_path):
            continue  # Dosya değil klasör olmalı
        
        # Görüntü ve maske hacimlerini oku
        image_volume = read_image_series(image_path)
        mask_volume = read_image_series(mask_path)

        # .nii.gz olarak kaydet
        image_nii = nib.Nifti1Image(image_volume.astype(np.int16), affine=np.eye(4))
        mask_nii = nib.Nifti1Image(mask_volume.astype(np.uint8), affine=np.eye(4))

        # Kayıt
        nib.save(image_nii, os.path.join(output_image_dir, f"{patient_id}.nii.gz"))
        nib.save(mask_nii, os.path.join(output_mask_dir, f"{patient_id}.nii.gz"))

        print(f"{patient_id} için .nii.gz dosyaları kaydedildi.")

# Kullanım
image_folder = r"C:\Users\ilgar\Desktop\Biyomedikal\7.Quiz\image"
mask_folder = r"C:\Users\ilgar\Desktop\Biyomedikal\7.Quiz\mask"

# Aynı klasöre yazılacaksa:
convert_to_nii_gz(image_folder, mask_folder, image_folder, mask_folder)
```
